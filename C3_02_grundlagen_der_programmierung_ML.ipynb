### Aufgabe 1

Nachfolgend wird an mehreren Beispielen die Deklaration, Initialisierung sowie Neu- und Wiederbelegung einer Variablen gezeigt. 

__Eure Aufgabe ist es, den Wert und den Datentyp der benannten Variablen am Ende des kleinen Programms zu bestimmen.__ 

Versucht zunächst, dies **ohne** Ausführung des Codes zu lösen, also nur durch zeilenweises Nachvollziehen des Programms.

Notiert eure Lösungen jeweils dort, wo steht: ```... eure Antwort hier ...```. Ändern könnt ihr die Texte durch einen Doppelklick auf die entsprechende Stelle. 

````python
number = 1
number = number + 5
number = number * 2
number = number % 12
print(number, type(number))
````

### Aufgabe 2

In der folgenden Tabelle sind verschiedene Werte vorgegeben. 

Anhand dieser Werte sollt ihr ausprobieren, wie sich verschiedene Typumwandlungen auswirken. Dazu könnt ihr die folgende Programmstruktur als Hilfsmittel nutzen. 

````python
wert_vorher = 1.9 # <-- Spalte 1 der Tabelle
print(wert_vorher, type(wert_vorher))

print('---')

wert_nachher = int(wert_vorher) # <-- an der Stelle muss immer die Typisierungsfunktion geändert werden
print(wert_nachher, type(wert_nachher))
````

**Vorgehensweise:**

* Ihr beginnt in der ersten Zeile der Tabelle. Dort steht in der Spalte ```Wert (vorher)``` der Wert ```1.9```. Im oberen Programm weist ihr der Variable `wert_vorher` genau diesen Wert zu. 

* Außerdem enthält die Tabelle die gewünschte Typisierungsfunktion. Diese wird in der vierten Codezeile im obigen Codeblock verwendet (der Kommentar gibt einen Hinweis). 

* Führt ihr das Programm aus, so erhaltet ihr zunächst den Wert und Datentyp **vor der Typisierung** und anschließend den Wert und Datentyp **nach der Typisierung**. 

* Tragt eure Ergebnisse in die Tabelle ein. Sollte es zu Besonderheiten kommen, wie direkt in der ersten Zeile der Tabelle, tragt diese in der letzten Spalte ein.

* Änderungen in der Tabelle könnt ihr vornehmen, indem ihr doppelt auf die Tabelle klickt.

### Aufgabe 3

__Implementiere ein Programm, das die folgenden Funktionalitäten erfüllt:__

* Die Nutzerin oder der Nutzer soll aufgefordert werden, eine Zahl einzugeben. 

* Diese Eingabe soll zunächst in einer Variablen gespeichert werden. 

* Aus der Eingabe soll die Quadratzahl berechnet werden. 

* Die Ausgabe soll in etwa so aussehen: 'Die Quadratzahl deiner eingegebenen Zahl `ZAHL` ist `ERGEBNIS`.' Die Platzhalter sollen durch die tatsächlich eingegebenen und berechneten Werte ersetzt werden. 

````python
# HIER IST PLATZ FÜR EURE LÖSUNG! :)

zahl = input('Gib eine Zahl ein: ')
zahl = float(zahl) # die Aufgabenstellung ist hier unkonkret, int wäre daher auch möglich, aber float ist flexibler
quadrat = zahl ** 2 
print(f'Die Quadratzahl deiner eingegebenen Zahl {zahl} ist {quadrat}.')
````

### Aufgabe 4

**Implementiere einen kleinen Taschenrechner für die Grundrechenarten:** 
Dazu sollen die Nutzerin oder der Nutzer zwei Zahlen $ a $ und $ b $ eingeben. Das Programm gibt dann die Summe $ a + b $, die Differenz $ a - b $, das Produkt $ a \cdot b $ und den Quotienten $ a : b $ aus.

````python
# HIER IST PLATZ FÜR EURE LÖSUNG! :)

a = input('Gib eine Zahl a ein: ')
b = input('Gib eine Zahl b ein: ')

a = float(a)
b = float(b)

sum = a + b
dif = a - b
prod = a * b
quot = a / b 

print(f'{a} + {b} = {sum}')
print(f'{a} - {b} = {dif}')
print(f'{a} * {b} = {prod}')
print(f'{a} / {b} = {quot}')
````

### Aufgabe 5

**Implementiere ein Programm, das Eingaben, Typumwandlungen und Ausgaben miteinander verbindet:**

* Frage nach dem Namen einer Person.
* Frage nach einer Lieblingszahl und wandle die Eingabe in eine ganze Zahl um.
* Berechne das Doppelte dieser Lieblingszahl.
* Gib einen vollständigen Satz aus, der den Namen, die Lieblingszahl und das Doppelte enthält.

Achte auf passende Variablennamen und darauf, dass Eingaben zunächst als Text gespeichert werden.

````python
# HIER IST PLATZ FÜR EURE LÖSUNG! :)

name = input('Gib deinen Namen ein: ')

lieblingszahl = input('Gib deine Lieblingszahl ein: ')
lieblingszahl = float(lieblingszahl)

doppelt = lieblingszahl * 2

print(f'Hallo {name}, deine Lieblingszahl ist {lieblingszahl}. Das Doppelte davon ist {doppelt}.')
````
